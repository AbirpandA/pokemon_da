```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 

# Load the dataset
data = pd.read_csv('pokemon2.csv')
```

```python
print(data[['name', 'type1', 'type2', 'attack','abilities','pokedex_number']].head(10))
```

```python
plt.figure(figsize=(10,6))
```

```python

# Identify weakness columns (those starting with 'against_')
against_cols = [col for col in data.columns if col.startswith('against_')]

# Create a new DataFrame with only the type and weakness columns
type_weaknesses = data[['type1'] + against_cols]

# Calculate the number of weaknesses for each type
# A weakness is where the 'against_' value is greater than 1
weakness_counts = type_weaknesses.set_index('type1').gt(1).sum(axis=1)

# Group by type and get the mean of the weakness counts
# This handles multiple Pokémon of the same type
weakest_type = weakness_counts.groupby('type1').mean().sort_values(ascending=False)

print("Pokemon types ranked by number of weaknesses:")
print(weakest_type)

# Get the weakest type
print(f"\nThe weakest type is: {weakest_type.idxmax()} with {weakest_type.max()} weaknesses.")
```
